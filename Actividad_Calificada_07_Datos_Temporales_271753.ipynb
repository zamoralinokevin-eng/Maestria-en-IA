### 2.3 Velocidad promedio (6 pts)

La velocidad promedio de un viaje, en millas por hora, es `distance / (duracion_min / 60)`.

1. Crea la columna `velocidad_mph` **solo para viajes con duración mayor que cero**; en los demás debe quedar como faltante (`NaN`). Sugerencia: calcula la velocidad para todos los viajes y conserva solo los valores válidos con `.where(condición)`, que deja `NaN` donde la condición es `False`.
2. Cuenta los viajes con velocidad mayor a 60 mph (velocidad poco plausible en tráfico urbano).

## Actividad 3. Componentes y patrones (25 puntos)

### 3.1 Extraer componentes (6 pts)

Crea las columnas:

- `hora`: hora de `pickup`.
- `dia_num`: día de la semana de `pickup` (lunes = 0).
- `dia_nombre`: nombre del día de `pickup`.
- `fin_semana`: `True` si es sábado o domingo.
- `fecha`: solo la fecha (sin hora) de `pickup`.
- `cruza_medianoche`: `True` si la **fecha** de `dropoff` es distinta de la de `pickup`.

Cuenta cuántos viajes cruzan la medianoche.

```python
#se crea la columna hora
taxis["hora"] = taxis["pickup"].dt.hour

#se crea la columna dia_num
taxis["dia_num"] = taxis["pickup"].dt.dayofweek

#se crea la columna dia_nombre
taxis["dia_nombre"] = taxis["pickup"].dt.day_name()

#se crea fin de semana si es sabado o domingo
taxis["fin_semana"] = taxis["dia_nombre"].isin(["Saturday", "Sunday"])

#se crea la columna fecha
taxis["fecha"] = taxis["pickup"].dt.date

#se crea la columna cruza_medianoche
taxis["cruza_medianoche"] = taxis["pickup"].dt.date != taxis["dropoff"].dt.date
print("Viajes que cruzan la medianoche:", taxis["cruza_medianoche"].sum())
```

### 3.2 Patrón por hora del día (7 pts)

Con `groupby('hora')`, calcula para cada hora: el **número de viajes** y la **mediana** de `duracion_min`. Indica la hora con más viajes y la hora con menos.

Usa la mediana (no la media) porque la duración tiene valores extremos.

```python
#se calcula para cada hora el numero de viajes y mediana
patron_por_hora = taxis.groupby("hora").agg(
    numero_viajes = ("pickup", "count"),
    mediana_duracion = ("duracion_min", "median")
)

#se indica la hora con mas viajes y la hora con menos viajes
hora_mas_viajes = patron_por_hora["numero_viajes"].idxmax()
hora_menos_viajes = patron_por_hora["numero_viajes"].idxmin()

#se imprimen valores, incluyendo el uso de la mediana
print("\nPatrón de numero de viajess por hora del día:")
print(patron_por_hora)
print(f"\nHora con mas viajes: {hora_mas_viajes}h con {patron_por_hora.loc[hora_mas_viajes, 'numero_viajes']} viajes.")
print(f"Hora con menos viajes: {hora_menos_viajes}h con {patron_por_hora.loc[hora_menos_viajes, 'numero_viajes']} viajes.")


```

### 3.3 Patrón por día de la semana (6 pts)

Calcula por día de la semana el **número de viajes** y la **mediana** de `duracion_min` y de `fare`. Presenta los días en orden de lunes a domingo (sugerencia: agrupa por `['dia_num', 'dia_nombre']`).

Toma en cuenta que marzo de 2019 no tiene el mismo número de cada día de la semana: calcula también el **promedio de viajes por fecha**, dividiendo entre el número de fechas distintas de ese día de la semana (usa la columna `fecha` y `.nunique()`).

```python
#se agrupan los datos  por dia_num y di_nombre
patron_por_dia = taxis.groupby(["dia_num", "dia_nombre"]).agg(
    numero_viajes = ("pickup", "count"),
    mediana_duracion = ("duracion_min", "median"),
    mediana_tarifa = ("fare", "median")
).reset_index()

#se calcula el promedio de viajes por fecha
fechas_por_dia = taxis.groupby(["dia_num", "dia_nombre"])["fecha"].nunique().reset_index(name='fechas_unicas')
patron_por_dia = patron_por_dia.merge(fechas_por_dia, on=["dia_num", "dia_nombre"])

#se calcula el promedio de viajes por fecha
patron_por_dia["promedio_viajes_por_fecha"] = patron_por_dia["numero_viajes"] / patron_por_dia["fechas_unicas"]

#se imprimer los resultados obtenidos
print("Análisis estandarizado por día de la semana")
print(patron_por_dia[['dia_nombre', 'numero_viajes', 'fechas_unicas', 'promedio_viajes_por_fecha', 'mediana_duracion', 'mediana_tarifa']])
```

## Actividad 4. Dataset limpio y comparación (20 puntos)

### 4.1 Dataset limpio (6 pts)

Crea `taxis_limpio` aplicando el criterio que definiste en 2.4. Imprime cuántos viajes se excluyeron por cada condición y en total, y verifica que `taxis_limpio` cumple todas las condiciones.

```python
#se aplican los filtros de la seccion 2.4, con condiciones por periodo, duracion y velocidad
condicion_periodo = (taxis["pickup"].dt.year == 2019) & (taxis["pickup"].dt.month == 3)
condicion_duracion = (taxis["duracion_min"] > 1)
condicion_velocidad = (taxis["velocidad_mph"] < 60)&(taxis["velocidad_mph"].notna())
taxis_limpio = taxis[condicion_periodo & condicion_duracion & condicion_velocidad]

#se cuantifican datos con los filtros creados
viajes_excluidos_periodo = len(taxis) - condicion_periodo.sum()
viajes_excluidos_duracion = len(taxis) - condicion_duracion.sum()
viajes_excluidos_velocidad = len(taxis[taxis["velocidad_mph"]>60])

#se crea el dataframe taxis_limpio
taxis_limpio = taxis[condicion_periodo & condicion_duracion & (taxis['velocidad_mph'] <= 60)].copy()
total_excluidos = len(taxis) - len(taxis_limpio)

print(f"Viajes excluidos por periodo (fuera de marzo 2019): {viajes_excluidos_periodo}")
print(f"Viajes excluidos por duración < 1 minuto: {viajes_excluidos_duracion}")
print(f"Viajes excluidos por velocidad > 60 mph: {viajes_excluidos_velocidad}")
print(f"Ttoal de registros excluidos: {total_excluidos}")
print(f"Dimensiones finales de 'taxis_limpio': {taxis_limpio.shape}")
```

### 4.2 Comparación entre periodos (6 pts)

Con `taxis_limpio`, compara los viajes que inician entre las **7:00 y las 9:59** en **días hábiles** contra los del **mismo horario en fin de semana**. Para cada grupo reporta: número de viajes, mediana de `duracion_min` y mediana de `velocidad_mph`.
