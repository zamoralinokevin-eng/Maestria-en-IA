## Actividad 3 — Alta cardinalidad (25 pts)

```python
# 3.1 — Calcula .nunique() para TODAS las columnas del dataset (no solo las categóricas)
# y la razón (valores únicos / total de filas) para cada una.
for col in df.columns:
  n_unicos = df[col].nunique()
  #se genera el ratio dividiendo los valores unicos para el total de filas.
  ratio = n_unicos / len(df)
  print(f"{col:12s}   {n_unicos:5d} valores unicos ({ratio:.1%} del total de filas)")

```

**3.3 — Agrupación "Top 10 + Otros"**

En clase agrupaste `country` de Netflix Titles en sus 10 categorías más frecuentes + `'Otros'`, reduciendo su cardinalidad. Aplica la misma técnica aquí sobre la columna de mayor cardinalidad que identificaste en 3.1 (pista: `.value_counts().head(10)`, luego `.where()` + `.isin()`, igual que en el notebook de clase).

```python
# Aplica el agrupamiento Top 10 + Otros sobre la columna de mayor cardinalidad
#se hace una lista con el top 10 de customerID
top10_customerID = df["customerID"].value_counts().head(10).index.tolist()
#se crea una nueva columna donde se reemplazan los valores fuera del top 10 en categoria "Otros"
df["customerID_agrupado"] = df["customerID"].where(df["customerID"].isin(top10_customerID), "Otros")
#se muestran los valores unicos de la columna generada
print(df["customerID_agrupado"].value_counts())

```

## Actividad 4 — Tipos de dato (30 pts)

```python
df['TotalCharges'].dtype
```

**4.1 — Investiga (responde en Markdown):**

`TotalCharges` contiene valores numéricos (montos en dólares), pero pandas la cargó como `object`, no como `float`. Investiga por qué — revisa si hay algún valor que no se vea como un número normal.

_Tu respuesta:_Cuando una columna que deberia tener valores decimales tiene cargados valores que a simple vista parecen float pero hay espcios en blanco antes o despues del valor, valores nulos, celdas vacias, caracteres invisibles, o separadores con el punto y la coma, Pandas lo interpreta como Object en vez de float. En el caso de la columna TotalCharges, existen valores vacios y de mas de tres cifras enteras que estan separados por la coma y luego al punto decimal. Ambos separadores juntos no correponden al formato numerico estandar que pandas pueda convertir a float y por eso cuando se lee el dtype aparece como Object

```python
# 4.2 — Corrige el tipo de TotalCharges.
# Pista: pd.to_numeric() con el parámetro errors= te puede ayudar a identificar
# o manejar los valores problemáticos que encontraste en 4.1.
df["TotalCharge"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
#se reemplazan los valores nulos por ceros
df["TotalCharge"] = df["TotalCharge"].fillna(0)
#se muestra el tipo de datos cambiado
print(df["TotalCharge"].dtype)

```

```python
# 4.3 — Convierte a category las columnas categóricas que, según lo que calculaste
# en la Actividad 3, tengan cardinalidad baja y valores fijos.
# Verifica con .dtypes que el cambio se aplicó correctamente.
columnas_a_cambiar = ["gender", "Partner", "Dependents", "PhoneService", "MultipleLines",
                     "InternetService", "OnlineSecurity", "OnlineBackup", "DeviceProtection",
                     "TechSupport", "StreamingTV", "StreamingMovies", "Contract", "PaperlessBilling",
                     "PaymentMethod", "Churn"]
#se hace en ciclo para cambiar cada columna de object a category
for col in columnas_a_cambiar:
  df[col] = df[col].astype("category")
  print(col, df[col].dtype)

```
