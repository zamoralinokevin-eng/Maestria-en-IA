### 1.1 Conteo de duplicados exactos (5 pts)

Calcula el número de filas marcadas como duplicadas en `hoteles` considerando **todas** las columnas, con `keep='first'`, `keep='last'` y `keep=False`. Imprime los tres resultados con un texto que explique qué significa cada uno.

```python
#Duplicado de Hoteles con keep='first'
duplicados_first = hoteles.duplicated(keep='first').sum()
print("Filas duplicadas con keep='first':",duplicados_first )
#Duplicado de Hoteles con keep='last'
duplicados_last = hoteles.duplicated(keep='last').sum()
print("Filas duplicadas con keep='last':", duplicados_last)
#Duplicado de Hoteles con keep=False
duplicados_false = hoteles.duplicated(keep=False).sum()
print("Filas duplicadas con keep=False:", duplicados_false)
```

### 1.2 Magnitud del problema (5 pts)

Con los resultados de 1.1, calcula e imprime:

- El **porcentaje** de filas sobrantes respecto al total.
- El número de **grupos repetidos** (combinaciones de valores que aparecen más de una vez).

```python
#Calculo del porcentaje de filas sobrantes
porcentaje_filas_sobrantes = duplicados_first / len(hoteles) * 100
print(f"El porcentaje de filas sobrantes es: {porcentaje_filas_sobrantes:.1f}%")

#Calculo del número de grupos repetidos opcion 1
conteo_grupos = hoteles.value_counts()
grupos_repetidos = conteo_grupos[conteo_grupos > 1].count()
print(f"El conteo de filas que aparecen mas de una vez (Opcion 1) son: {grupos_repetidos}")

#Calculo del número de grupos repetidos opcion 2
repetidos_separados = hoteles.duplicated(keep='first').sum()
repetidos_juntos = hoteles.duplicated(keep=False).sum()
grupos_opcion2 = repetidos_juntos - repetidos_separados
print(f"El conteo de filas que aparecen mas de una vez (Opcion 2) son: {grupos_opcion2}")


```

### 1.3 Duplicados por hotel (5 pts)

Determina **cuántas** filas sobrantes (`keep='first'`) hay en cada hotel y qué **porcentaje** representan respecto al total de reservas de ese hotel.

Sugerencia: crea una copia de `hoteles`, agrega una columna booleana `es_duplicado` con el resultado de `duplicated()` y agrupa por `hotel`.

```python
#Se crea un copia de hoteles
hoteles_duplicados = hoteles.copy()

#se agrega la columna booleana "es_duplicado" con duplicated()
hoteles_duplicados['es_duplicado'] = hoteles_duplicados.duplicated(keep='first')

#se agrupa por columna "hotel", se muestran los valores duplicados y totales en columnas independientes
hoteles_agrupados = hoteles_duplicados.groupby('hotel')['es_duplicado'].agg(duplicados='sum',total='count').reset_index()

#se calcula el porcentaje con duplicados en el grupo/total del grupo
hoteles_agrupados['porcentaje'] = hoteles_agrupados['duplicados'] / hoteles_agrupados['total'] * 100
hoteles_agrupados
```

## Actividad 3. Impacto en el análisis (25 puntos)

### 3.1 Versión sin duplicados (5 pts)

Crea `hoteles_limpio` eliminando los duplicados exactos, conservando la primera aparición y **renumerando el índice**. Imprime el número de filas antes y después y cuántas se eliminaron.

```python
#se crea hoteles_limpio eliminando los duplicados. Se renumera el indice
hoteles_limpio = hoteles.drop_duplicates(keep='first').reset_index(drop=True)

#se imprime el nuero de filas antes, despues y cuantas se eliminaron
filas_antes = len(hoteles)
filas_despues = len(hoteles_limpio)
filas_eliminadas = filas_antes - filas_despues
print(f"Filas antes: {filas_antes}")
print(f"Filas despues: {filas_despues}")
print(f"Filas eliminadas: {filas_eliminadas}")
```

### 3.2 Métricas por hotel (8 pts)

Para `hoteles` y para `hoteles_limpio`, calcula por hotel con `groupby()`:

- Número de reservas (`.size()`).
- Tasa de cancelación (media de `is_canceled`).
- Tarifa diaria promedio (media de `adr`).

Nota: para calcular la media de varias columnas a la vez, selecciona una lista de columnas: `df.groupby('hotel')[['col1', 'col2']].mean()`.

Guarda las medias en dos `DataFrame`, `resumen_original` y `resumen_limpio`, con `hotel` como columna (usa `reset_index()`).

```python
#se calculan metricas para el hoteles (dataset de inicio)
resumen_original = hoteles.groupby('hotel').agg(
    numero_reservas=('hotel', 'size'),
    tasa_cancelacion=('is_canceled', 'mean'),
    tarifa_diaria_promedio=('adr', 'mean')
).reset_index()

#se calculan metricas para el hotel sin duplicados
resumen_limpio = hoteles_limpio.groupby('hotel').agg(
    numero_reservas=('hotel', 'size'),
    tasa_cancelacion=('is_canceled', 'mean'),
    tarifa_diaria_promedio=('adr', 'mean')
).reset_index()

#se guardaron las medias en los dataframes de arriba
print("Resumen original:\n", resumen_original)
print("\nResumen limpio:\n", resumen_limpio)
```

### 3.3 Tabla comparativa (6 pts)

Agrega a cada resumen una columna `version` (`'original'` y `'sin duplicados'`) y únelos con `pd.concat()` en un `DataFrame` llamado `comparacion`, con índice renumerado y ordenado por `hotel`, de modo que las dos versiones de cada hotel queden juntas.

```python
#se añaden las etiquetas de las versiones de los dataframes
resumen_original['version'] = 'original'
resumen_limpio['version'] = 'sin duplicados'

#se unen los dataframes con pd.concat() en uno solo
comparacion = pd.concat([resumen_original, resumen_limpio], ignore_index=True)

#se ordenan los valores por hotel
comparacion = comparacion.sort_values(by='hotel').reset_index(drop=True)
comparacion
```
